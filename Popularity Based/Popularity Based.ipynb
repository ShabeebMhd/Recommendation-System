```python
import pandas as pd
import numpy as np
```

```python
# Movie metadata
i_cols = ['movie id', 'movie title', 'release date', 'video release date', 'IMDb URL', 'unknown', 'Action', 'Adventure',
    'Animation', "Children's", 'Comedy', 'Crime', 'Documentary', 'Drama', 'Fantasy',
    'Film-Noir', 'Horror', 'Musical', 'Mystery', 'Romance', 'Sci-Fi', 'Thriller', 'War', 'Western']
items = pd.read_csv('ml-100k/u.item', sep='|', names=i_cols, encoding='latin-1')
```

```python
# Ratings data — this is what actually tells you how many times each movie was watched
r_cols = ['user id', 'movie id', 'rating', 'timestamp']
ratings = pd.read_csv('ml-100k/u.data', sep='\t', names=r_cols, encoding='latin-1')

# Count ratings per movie = proxy for "watched"
watch_counts = ratings.groupby('movie id').size().reset_index(name='watch_count')

# Attach watch counts to the movie table
movies = items.merge(watch_counts, on='movie id', how='left')
movies['watch_count'] = movies['watch_count'].fillna(0).astype(int)
```

```python
genre_cols = ['Action', 'Adventure', 'Animation', "Children's", 'Comedy',
              'Crime', 'Documentary', 'Drama', 'Fantasy', 'Film-Noir', 'Horror',
              'Musical', 'Mystery', 'Romance', 'Sci-Fi', 'Thriller', 'War', 'Western']

for genre in genre_cols:
    genre_movies = movies[movies[genre] == 1]
    top5 = (genre_movies.sort_values('watch_count', ascending=False)
                         .head(5)[['movie title', 'watch_count']]
                         .reset_index(drop=True))
    top5.index += 1  # rank starting at 1

    print(f"\n{genre}")
    print("-" * (len(genre) + 2))
    for rank, row in top5.iterrows():
        print(f"  {rank}. {row['movie title']:<40} {row['watch_count']:>4} watches")
```
